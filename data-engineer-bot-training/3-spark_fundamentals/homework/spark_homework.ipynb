```python
# import libraries and create spark session
from pyspark.sql import SparkSession
from pyspark.sql.functions import expr, col
from pyspark.sql.functions import broadcast

spark = SparkSession.builder.appName("SparkHomework").getOrCreate()
```

```python
# read the datasets 
df_matches = spark.read.option("header", "true").csv("/home/iceberg/data/matches.csv")

df_maps = spark.read.option("header", "true").csv("/home/iceberg/data/maps.csv") 

df_match_details = spark.read.option("header", "true").csv("/home/iceberg/data/match_details.csv")

df_medals_matches_players = spark.read.option("header", "true").csv("home/iceberg/data/medals_matches_players.csv")
```

```python
# Disable automatic broadcast join 
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "-1")
```

```python
# broadcast join medals and maps - to do
df_matches_with_maps = df_matches.join(broadcast(df_maps), on="mapid", how="inner")

# to verify if broadcast join happen
df_matches_with_maps.explain()
```

```python
# bucket join match_details, matches and medal_matches_players on match_id with 16 buckts - to do

# store the data into tables to perform bucket join 
df_matches.write \
    .mode("override") \
    .bucketBy(16, "match_id") \
    .saveAsTable("bootcamp.matches_bucketed")

df_match_details.write \
    .mode("override") \
    .bucketBy(16, "match_id") \
    .saveAsTable("bootcamp.match_details_bucketed")

df_medals_matches_players.write \
    .mode("overriede") \
    .bucketBy(16, "match_id") \
    .saveAsTable("bootcamp.medals_matches_players_bucketed")
```
